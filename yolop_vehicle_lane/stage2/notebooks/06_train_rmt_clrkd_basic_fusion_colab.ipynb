```python
import os
DRIVE_ECOCAR = '/content/drive/MyDrive/EcoCAR'
DRIVE_DATASETS = os.path.join(DRIVE_ECOCAR, 'datasets')
DRIVE_TRAINING_RUNS = os.path.join(DRIVE_ECOCAR, 'training_runs')
os.makedirs(DRIVE_TRAINING_RUNS, exist_ok=True)
CURVE_TAR = os.path.join(DRIVE_DATASETS, 'bdd100k_clrkd_curve.tar')
assert os.path.exists(CURVE_TAR), CURVE_TAR
DET_LABEL_ROOT = '/content/bdd100k_vehicle5/labels'
if not os.path.exists(DET_LABEL_ROOT):
    print('Warning: detection labels not found at', DET_LABEL_ROOT)
    print('Run the Stage-1 dataset preparation first, or pass another --det-label-root path to the script.')
else:
    print('Detection labels:', DET_LABEL_ROOT)
```

```python
import os, subprocess
cmd = [
    'python', 'stage2/scripts/06_train_rmt_clrkd_basic_fusion.py',
    '--config', 'stage2/configs/rmt_clrkd_basic_fusion.yaml',
    '--curve-tar', CURVE_TAR,
    '--curve-root', '/content/bdd100k_clrkd_curve',
    '--work-dir', '/content/stage2_rmt_clrkd_basic_fusion',
    '--output-tar', os.path.join(DRIVE_TRAINING_RUNS, 'stage2_rmt_clrkd_basic_fusion.tar'),
]
if os.path.exists(DET_LABEL_ROOT):
    cmd += ['--det-label-root', DET_LABEL_ROOT]
print(' '.join(cmd))
subprocess.check_call(cmd)
```
