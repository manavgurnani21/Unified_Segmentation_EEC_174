# Stage 2 Notebook 66 - Exp2KKK Anchor + cls_sep + IoU-priority matching + full 70K + 12ep

**Cleaner matching for the cls_sep recipe.** NB62's matching cost weights were `match_cost_point=5.0, match_cost_iou=2.0` -- point distance heavily favored over IoU. This may match priors that are coordinate-close but have LOW IoU with the GT, producing noisy cls supervision.

Exp2KKK shifts cost weighting to favor IoU: `match_cost_point: 5.0 -> 2.0`, `match_cost_iou: 2.0 -> 5.0`. Also bumps `w_iou: 2.0 -> 3.0` to reinforce that the matched priors actually have high LineIoU with their assigned GT.

Hypothesis: cleaner positive matches (higher IoU on matched priors) gives cls a more discriminative signal. Combined with cls_sep's disjoint feature pathway, gap should exceed NB62's 0.045.
