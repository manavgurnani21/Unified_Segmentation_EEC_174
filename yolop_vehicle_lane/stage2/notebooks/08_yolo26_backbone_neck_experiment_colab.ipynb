```python
import os, pathlib
CANDIDATE_ROOTS = [
    pathlib.Path('/content/drive/MyDrive/EcoCAR'),
    pathlib.Path('/content/drive/MyDrive'),
    pathlib.Path('/content'),
]
hits = []
for base in CANDIDATE_ROOTS:
    if not base.exists():
        continue
    for child in base.iterdir():
        name = child.name.lower()
        if child.is_dir() and ('yolo26' in name or 'yolov26' in name):
            hits.append(child)
print('YOLO26 candidates:')
for h in hits:
    print(' ', h)
if not hits:
    raise FileNotFoundError('No YOLO26 project folder found. Put it at the GitHub workspace top level or update CANDIDATE_ROOTS.')
YOLO26_DIR = str(hits[0])
print('selected:', YOLO26_DIR)
```
